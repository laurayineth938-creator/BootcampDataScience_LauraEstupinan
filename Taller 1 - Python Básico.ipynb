```python
# Funciones y Manejo de Errores


def calcular_ingreso(precio, cantidad):
    try: 
    
     if not isinstance(precio, (int, float)) or not isinstance(cantidad, int):

        raise TypeError("Los valores deben ser numéricos")
     else:
      calcular_ingreso = precio * cantidad
 
      return calcular_ingreso
    
    
    except TypeError:
        print("Error: Los valores deben ser numéricos")

        return 0
    
```
