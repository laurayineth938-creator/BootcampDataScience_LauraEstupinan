# usamos import request para cargar en Python las herramientas que necesitamos para hacee peticiones HTTP a servidores en internet y consumir datos de una API.


```python

import pandas as pd
import requests
```

# Definimos las variables de configuracion de la API de OpenWeatherMap


```python

api_key = "257678fbbbb0a5076959b98a4f859b49"
ciudades = [
    "Ciudad de Panama",
    "Colon",
    "David",
    "Chitre",
    "Santiago",
    "Bocas del Toro",
    "La Chorrera",
]
lista_datos = []
for ciudad in ciudades:

# construimos la URL de la API de OpenWeatherMap con la ciudad y la clave de API

  url = f"https://api.openweathermap.org/data/2.5/weather?q={ciudad}&appid={api_key}&units=metric&lang=es"

# enviamos la solicitud GET a la API y obtenemos la respuesta en formato JSON
  respuesta = requests.get(url)
  datos = respuesta.json()

# verificamos si la respuesta fue exitosa (código 200) y extraemos los datos relevantes para cada ciudad

  if datos.get("cod") == 200:
    lista_datos.append({
        "Ciudad": datos["name"],
        "Pais": datos["sys"]["country"],
        "Temperatura_C": datos["main"]["temp"],
        "Sensacion_Termica_C": datos["main"]["feels_like"],
        "Temp_Min_C": datos["main"]["temp_min"],
        "Temp_Max_C": datos["main"]["temp_max"],
        "Humedad_Pct": datos["main"]["humidity"],
        "Presion_hPa": datos["main"]["pressure"],
        "Viento_m_s": datos["wind"]["speed"],
        "Nubosidad_Pct": datos["clouds"]["all"],
        "Estado_Clima": datos["weather"][0]["description"].capitalize(),
    })

```

# Creamos un DataFrame de pandas con los datos obtenidos y lo guardamos en un archivo CSV llamado "clima_panama.csv"
# usamos encoding="utf-8-sig" para que los caracteres especiales se muestren correctamente en Power BI.
# (-sig) Agrega una marca invisible llamada BOM (Byte Order Mark) al inicio del archivo.para que interprete bien el texto ya que tiene caracteres especiales como acentos y eñes.

```python

df = pd.DataFrame(lista_datos)
df.to_csv("clima_panama.csv", index=False, encoding="utf-8-sig")


print (pd.DataFrame(lista_datos))
print("¡Archivo 'clima_panama.csv' guardado con éxito!")
```
